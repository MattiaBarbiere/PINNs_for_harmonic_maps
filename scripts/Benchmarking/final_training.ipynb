## Eigenfunction with other frequencies

I will try to solve the following poisson problem
\begin{equation}
\Delta u = f(x, y)
\end{equation}

with boundary conditions:

\begin{equation}
u(0, y) = u(1, y) = u(x, 0) = u(x, 1) = 0
\end{equation}

where

\begin{equation}
f(x, y) = -\sin\left(\frac{\pi x}{a}\right) \sin\left(\frac{\pi y}{b}\right)
\end{equation}

The analytical solution to this is
\begin{equation}
u(x, y) = \frac{1}{2 \pi^2} \sin\left(\frac{\pi x}{a}\right) \sin\left(\frac{\pi y}{b}\right)
\end{equation}

## Gaussian Bump

$$A(x,y):H(x,y) = f(x,y)$$

Where
\begin{equation}
A(x, y) = \begin{pmatrix}
1 & 0 \\
0 & 1 \\
\end{pmatrix}
\end{equation}

and
\begin{equation}
H(x, y) = \begin{pmatrix}
\frac{\partial^2 u}{\partial x^2} & \frac{\partial^2 u}{\partial x \partial y} \\
\frac{\partial^2 u}{\partial y \partial x} & \frac{\partial^2 h}{\partial y^2}
\end{pmatrix}
\end{equation}


With 

\begin{equation}
f(x, y) = \frac{1}{\sigma_x \sigma_y \sqrt{2 \pi}} \exp\left\{\frac{-(x - \mu_x)^2}{2 \sigma_x^2} - \frac{(y - \mu_y)^2}{2 \sigma_y^2} \right\}
\end{equation}

### $\mu_x$ = 0.25, $\mu_y$ = 0.25, $\sigma_x$ = 0.1, $\sigma_y$ = 0.25

```python
path = f"../../simulation_studies/final_training/outputs/gaussian_bump_NonDF_mu_x_{0.25}_mu_y_{0.25}_std_x_{0.1}_std_y_{0.25}"
compare_model_to_benchmark(path, print_PDE=True)
plot_errors_from_path(path)
```

### $\mu_x$ = 0.75, $\mu_y$ = 0.75, $\sigma_x$ = 0.25, $\sigma_y$ = 0.25

\end{equation}

and
\begin{equation}

\end{pmatrix}
\end{equation}
