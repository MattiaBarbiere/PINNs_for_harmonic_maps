# Quarter Annulus

The harmonic map problem that we need to solve is

$$ A(D u):H(u_i)= 0 \quad \text{ for } \quad i \in \{1,2\}$$

with
$$ A(D u) = \begin{pmatrix} (\partial_{x_2} u) \cdot (\partial_{x_2} u) & - (\partial_{x_1} u) \cdot (\partial_{x_2} u) \\ - (\partial_{x_1} u) \cdot (\partial_{x_2} u) & (\partial_{x_2} u) \cdot (\partial_{x_2} u) \end{pmatrix} $$
with the boundary condition of 
$$
g(x_1, x_2) = \begin{pmatrix} (x_1 + 1) \cos(x_2 (0.5 \pi)) \\  (x_1 + 1) \sin(x_2 (0.5 \pi))\end{pmatrix} \quad (x_1, x_2) \in \partial([0,1]^2)

$$

## The boundary condition

```python
from hmpinn.plotting.plotting_harmonic_maps import plot_grid_comparison, plot_boundary_comparison
from hmpinn.PDEs.harmonic_maps.quarter_annulus import quarter_annulus_BC
```

```python
plot_boundary_comparison(quarter_annulus_BC, title="Quarter annulus boundary")
```

## Losses

```python
path = "../simulation_studies/harmonic_maps/outputs/quarter_annulus_hm"
plot_errors_from_path(path)
```

```python
hm, _, _, _, _ = load_model(path)
hm.eval()

print(hm)
```

```python
plot_boundary_comparison(hm, title="Quarter annulus boundary")
```

```python
for n in grid_sizes:
    plot_grid_comparison(hm, grid_size_x=n, grid_size_y=n, title=f"Quarter annulus grid with n ={n}")
```

# Sine Boundaries

The harmonic map problem that we need to solve is

$$ A(D u):H(u_i)= 0 \quad \text{ for } \quad i \in \{1,2\}$$

with
$$ A(D u) = \begin{pmatrix} (\partial_{x_2} u) \cdot (\partial_{x_2} u) & - (\partial_{x_1} u) \cdot (\partial_{x_2} u) \\ - (\partial_{x_1} u) \cdot (\partial_{x_2} u) & (\partial_{x_2} u) \cdot (\partial_{x_2} u) \end{pmatrix} $$
with the boundary condition of 
$$
g(x_1, x_2) = \begin{pmatrix} (1 - c   \sin( f_x   \pi   x_2))   (2x_1 - 1) \\
 (1 - c   \sin( f_y   \pi   x_1))   (2x_2 - 1) \end{pmatrix} \quad (x_1, x_2) \in \partial([0,1]^2)
$$

where $c, f_x$ and $f_y$ are parameters that can be varied.

```python
from hmpinn.PDEs.harmonic_maps.sin_boundaries import sin_boundaries_BC
```

## Sine Boundaries $f_x = 1$ and $f_y = 1$

```python
f_x = 1
f_y = 1
path = f"../simulation_studies/harmonic_maps/outputs/sin_boundaries_hm_f_x_{f_x}_f_y_{f_y}"
```

## The boundary condition

```python
plot_boundary_comparison(sin_boundaries_BC, frequency_x=f_x, frequency_y=f_y,
                 title=f"Sin boundaries f_x = {f_x} and f_y = {f_y}")
```

### Losses

```python
plot_errors_from_path(path)
```

```python
hm, _, _, _, _ = load_model(path)
hm.eval()

print(hm)
```

```python
for n in grid_sizes:
    plot_grid_comparison(hm, grid_size_x=n, grid_size_y=n, title=f"Sin boundaries grid f_x = {f_x} and f_y = {f_y}",)
```
